**<center>list practices</center>**

```python
list=[20,30,40,50,60]
print(list)
print(type(list))
print(list[0])
list.insert(2,35)
print(list)
list.append(70)
print(list)
```

```python
list=['apple','banana','cherry']
list.append('orange')
print(list)
list.insert(1,'kiwi')
print(list)
list.append('grape')
print(list)
list.pop()
print(list)
list.pop(3)
print(list)
list.remove('banana')
print(list)
list.extend(['pommegranate','watermelon','mango'])
print(list)
list.append(['guava','blueberry'])
print(list)
```

```python
fruits=['apple', 'kiwi', 'banana', 'cherry', 'orange', 'grape']
for fruit in fruits:
    print(fruit)
```

```python
list=[20,30,40,50,60]
print(list)
sqrs=[]
for num in list:
    sqrs.append(num**2)
print(sqrs)
```

```python
list=[20,30,40,50,60]
result=[num*3 for num in list]
print(result)
```

```python
list=[15,26,33,41,25,36,57,88]
print(list)
result=[("even", num) if num%2==0 else ("odd", num) for num in list]
result1=[x for x in list if x%2==0]
print(result)
print(f"even numbers: {result1}")
```

```python
numbers = [10, 20, 30, 40, 50, 60]
print(numbers[:])
print(numbers[::])
print(numbers[::-1])
print(numbers[2:])
print(numbers[:4])
print(numbers[1:4])
print(numbers[::2])
print(numbers[5:2:-1])
print(numbers[1:5:2])
```
