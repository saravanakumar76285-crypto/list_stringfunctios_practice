```python
word = "Python Programming"
vowels = 'aeiou'
vowl = 0
consonent = 0

for char in word.lower():
    if char in vowels:
        vowl += 1
    elif char.isalpha():
        consonent += 1

print("Vowels:", vowl)
print("Consonants:", consonent)
```

```python
text = "Python123is45easy"
digits = 0
for char in text:
    if char.isdigit():
        digits += 1

print("Digits:", digits)
```

```python
text = "PyTHon ProGRAMming 123"
uppercase = 0
lowercase = 0
for char in text:
    if char.isupper():
        uppercase += 1
    elif char.islower():
        lowercase += 1
print("Uppercase letters:", uppercase)
print("Lowercase letters:", lowercase)
```
